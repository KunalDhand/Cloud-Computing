## Part I: Randomized Sales Data Generation

```python
import numpy as np
import pandas as pd

roll_number = 50  # change this
np.random.seed(roll_number)

months = 12
categories = 4

sales_data = np.random.randint(1000, 5001, size=(months, categories))
sales_data
```

```python
columns = ["Electronics", "Clothing", "Home & Kitchen", "Sports"]
months_index = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]

df = pd.DataFrame(sales_data, columns=columns, index=months_index)
df
```

## Part II: Data Manipulation

```python
df.head()
df.describe()
```

```python
total_category = df.sum()
total_month = df.sum(axis=1)

total_category, total_month
```

```python
growth = df.pct_change()
growth
```

```python
df["Total Sales"] = df.sum(axis=1)
df["Growth Rate"] = df["Total Sales"].pct_change() * 100
df
```

```python
if roll_number % 2 == 0:
    df["Electronics"] *= 0.9
else:
    df["Clothing"] *= 0.85

df
```

## Part III: Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns

df[columns].plot()
plt.title("Monthly Sales Trends")
plt.show()
```

```python
sns.boxplot(data=df[columns])
plt.title("Sales Distribution")
plt.show()
```

## Q2 Array Operations

```python
array = np.array([[1, -2, 3],[-4, 5, -6]])

abs_array = np.abs(array)

percentiles = np.percentile(array.flatten(), [25,50,75])
mean = np.mean(array)
median = np.median(array)
std = np.std(array)

abs_array, percentiles, mean, median, std
```

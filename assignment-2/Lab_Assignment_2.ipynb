## 3. Group similar elements

```python
from collections import defaultdict

matrix = [[1,2,2],[3,1,3],[2,1,3]]

group = defaultdict(list)

for i in range(len(matrix)):
    for j in range(len(matrix[0])):
        group[matrix[i][j]].append((i,j))

print(dict(group))
```

## 5. n x n matrix with even sum of opposite corners

```python
import numpy as np

n = 4
matrix = np.random.randint(0, 10, (n,n))

def check(matrix):
    n = len(matrix)
    for i in range(n-1):
        for j in range(n-1):
            sub = matrix[i:i+2, j:j+2]
            if (sub[0,0] + sub[1,1]) % 2 != 0:
                return False
    return True

print(matrix)
print("Condition satisfied:", check(matrix))
```

## 6. Row-wise addition in tuple matrix

```python
matrix = [(1,2,3),(4,5,6),(7,8,9)]

result = [sum(row) for row in matrix]
print(result)
```
